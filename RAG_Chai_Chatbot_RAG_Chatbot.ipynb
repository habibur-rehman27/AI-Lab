```python
class TrafficAgent:
    def __init__(self, light):
        self.light = light

    def Light(self):
        print("Traffic light:", self.light)


        if self.light == "Red":
            self.action("Red")
        elif self.light == "yellow":
            self.action("Wait")
        else:
            self.action("Go")

    def action(self, action):
        print("Agent reply:", action)


agent = TrafficAgent("RED")
agent.Light()

agent =TrafficAgent("YELLOW")
agent.Light()

agent = TrafficAgent("GREEN")
agent.Light()

```

```python
class VacuumAgent:
    def __init__(self):
        self.memory = []

    def move(self, room):
        self.memory.append(room)

    def show(self):
        print("Visited:", self.memory)


robot = VacuumAgent()

robot.move("kitchen")
robot.move("bedroom")
robot.move("lounge")

robot.show()
```

```python

position = 0
goal = 5

for i in range(position, goal):
    position = position + 1
    print("Robot is now at position", position)

print("Goal reached!")
```

```python

scores = {"left": 0, "right": 0}

feedback = [("left", -1), ("right", 1), ("right", 1), ("left", -1), ("right", 1)]

for action, reward in feedback:
    scores[action] = scores[action] + reward

print("Scores:", scores)

if scores["left"] > scores["right"]:
    print("The agent learned:", "left")
else:
    print("The agent learned:", "right")
```

```python

routes = {"Route A": 6, "Route B": 9, "Route C": 4}

best_route = max(routes, key=routes.get)

print("Best route:", best_route)
print("Utility:", routes[best_route])

```
